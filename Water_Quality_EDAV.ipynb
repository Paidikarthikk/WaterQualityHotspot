## Problem Overview

Water quality varies across different states, rivers, and sampling locations.
This project analyzes important water quality parameters to understand
pollution patterns and identify potential pollution hotspots.

The analysis focuses on parameters such as:

- Temperature
- Dissolved Oxygen (DO)
- pH
- Conductivity
- Biochemical Oxygen Demand (BOD)
- Nitrate and Nitrite
- Fecal Coliform
- Total Coliform

The project involves data loading, data cleaning, exploratory data analysis,
statistical analysis, missing-value treatment, visualization, and
multi-parameter hotspot identification.

## Dataset

The dataset is obtained from the Government of India's Data.gov.in
water-quality monitoring resources.

**Dataset:** Water Quality in India - 2013

**Source:** [Data.gov.in](https://data.gov.in/resources/water-quality-monitoring-india)

The collected data consists of multiple CSV files containing water-quality
measurements from different monitoring locations.

```python
from pathlib import Path

raw_data_path = Path("raw_data")

csv_files = list(raw_data_path.glob("*.csv"))

print("Number of CSV files found:", len(csv_files))
```

```python
# Read the first CSV file
df_test = pd.read_csv(csv_files[0], encoding="cp1252")

print("File loaded successfully!")
print("File name:", csv_files[0].name)
print("Shape:", df_test.shape)
```

```python
# Count missing values in each column
missing_values = df_test.isnull().sum()

print(missing_values)
```

```python
# Create an empty list to store each CSV DataFrame
all_data = []

# Read every CSV file
for file in csv_files:
    df = pd.read_csv(file, encoding="cp1252")
    
    # Store the source file name
    df["SOURCE_FILE"] = file.name
    
    # Add the DataFrame to our list
    all_data.append(df)

print("Files loaded:", len(all_data))   
```

```python
# Combine all 36 DataFrames into one master DataFrame
master_df = pd.concat(all_data, ignore_index=True)

print("Master dataset created!")
print("Rows:", master_df.shape[0])
print("Columns:", master_df.shape[1])
```
